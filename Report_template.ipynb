## 2. Citations

Like in any scientific work, it is essential that you correctly cite other work you used in the preparation of the report. The reader should be able to get a clear picture of how you came to the conclusions in your work and what sources were used to arrive at those conclusions.
For example, if you benefited from a textbook with title "Everything about FEM" by Dr. Donald Duck you can say something like "A valuable source for the material in Chapter 2 of this report is the textbook by D. By [1] you refer to a particular entry in your list of references at the end of the report, in this case a book published in 2003 by Cork University Press. If you refer to a recent article by Prof. Minnie Mouse and her coworkers you might say "As shown in [2] this calculation can be done in less than 5 years on a fast computer.". It is also important to cite at the right point, and make clear the extend of what you have used from your cited work. Any material, which is not connected to a cited source, would be assumed to be your original work.

Note that even if you cite a source you are not allowed to copy anything verbatim, unless you clearly mark your copy as word-for-word copy through quotation marks. ## 3. Figures

All figures need to have a caption with a figure number, so that you can refer to them from your main text. The purpose of the caption is to explain the technical content of the figure, i.e. what each
line means, maybe which equation or program was used to produce the figure, what parameter values were used, etc. The caption must not interpret or discuss the consequences of the figure. This is the task
of the main text. In the main text you refer to the figure through its label, for example, by writing "As shown in Fig. 1 a circle is significantly more red than a rectangle". It is not allowed to show figures without referring to them in the main text.

### Functions

Let us consider again our Fibonacci code snippet

    fiblist = [0,1]
    while len(fiblist) < 15:
        fiblist.append( fiblist[-1] + fiblist[-2])
    print(fiblist)
which prints a list of the first 15 Fibonacci numbers. If we want to calculate the first 20 numbers, we would need to change the `15` into `20`, but this does not seem very elegant. Instead let us define a function:

```python
def fibprint(n):
    fiblist = [0,1]
    while len(fiblist) < n:
        fiblist.append( fiblist[-1] + fiblist[-2])
    print(fiblist)
```
